```python
LAST_NAME = "Aragon"
STUDENT_ID = "TUPM-26-1834"
FAVORITE_ARTIST = "YOASOBI"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = 0
for char in FAVORITE_ARTIST.upper():
    if char.isalpha():
        value = ord(char) - ord('A') + 1
        ARTIST_SUM += value

print(f"LAST_NAME:       {LAST_NAME} (Length: {NAME_LENGTH})")
print(f"STUDENT_ID:      {STUDENT_ID} (Seed Digit: {SEED_DIGIT}, Sum: {ID_SUM})")
print(f"FAVORITE_ARTIST:  {FAVORITE_ARTIST} (Length: {ARTIST_LENGTH}, Alphabetical Sum: {ARTIST_SUM})")
```

```python
original_dataset = []
current_value = ID_SUM + 100  
for i in range(12):
    original_dataset.append(current_value)
    current_value += NAME_LENGTH

rotation_point = SEED_DIGIT % len(original_dataset)
if rotation_point == 0:
    rotation_point = 3  

front_part = original_dataset[0:rotation_point]
back_part = original_dataset[rotation_point:]
rotated_dataset = back_part + front_part

target_incident = ID_SUM + ARTIST_SUM + 100 

print(f"Original Sorted Dataset: {original_dataset}")
print(f"Rotation Point Index:    {rotation_point}")
print(f"Rotated Dataset:         {rotated_dataset}")
print(f"Target Incident ID:      {target_incident}")
```

```python
binary_comparisons = 0
binary_position = -1

low = 0
high = len(rotated_dataset) - 1

while low <= high:
    binary_comparisons += 1
    mid = (low + high) // 2
    mid_val = rotated_dataset[mid]
    
    if mid_val == target_incident:
        binary_position = mid
        break
        
    if rotated_dataset[low] <= mid_val:
        if rotated_dataset[low] <= target_incident < mid_val:
            high = mid - 1
        else:
            low = mid + 1
            
    else:
        if mid_val < target_incident <= rotated_dataset[high]:
            low = mid + 1
        else:
            high = mid - 1

if binary_position != -1:
    print(f"Target found at index position: {binary_position}")
else:
    print("Target incident NOT found in rotated system.")
print(f"Total Binary Search Comparisons: {binary_comparisons}")
```

```python
linear_comparisons = 0
linear_position = -1

for index in range(len(rotated_dataset)):
    linear_comparisons += 1
    if rotated_dataset[index] == target_incident:
        linear_position = index
        break

if linear_position != -1:
    print(f"Linear Search found target at index: {linear_position}")
else:
    print("Linear Search confirmed: Target NOT found.")
print(f"Total Linear Search Comparisons: {linear_comparisons}")
```

```python
print("\n1. Linear Search:")
print("   - Time Complexity:  O(n) - Must scan every item in worst-case.")
print("   - Space Complexity: O(1) - Uses a single index memory counter.")

print("\n2. Rotated Binary Search:")
print("   - Time Complexity:  O(log n) - Continuously eliminates half the data tracks.")
print("   - Space Complexity: O(1) - Only allocates low, mid, and high pointers.")
```
