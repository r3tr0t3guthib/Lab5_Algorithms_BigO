```python
import random
import time

LAST_NAME = "ARAGON"
STUDENT_ID = "TUPM-26-1834"
FAVORITE_ARTIST = "YOASOBI"

ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())              
NAME_LENGTH = len(LAST_NAME)      
ARTIST_LENGTH = len(FAVORITE_ARTIST) 

base_orders = []
current_id = ID_SUM + ARTIST_LENGTH + 500  
for i in range(10): 
    base_orders.append(current_id)
    current_id += NAME_LENGTH

sorted_input = list(base_orders)

reverse_input = list(base_orders)[::-1]

random.seed(42) 
random_input = list(base_orders)
random.shuffle(random_input)

print(f"Generated Base Order Dataset: {base_orders}")
print(f"1. Sorted Input Array:       {sorted_input}")
print(f"2. Reverse Input Array:      {reverse_input}")
print(f"3. Random Input Array:       {random_input}")
```

```python

def run_bubble_sort(input_array):
    arr = list(input_array)  
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                # Swap entries
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3  
    end_time = time.perf_counter()
    
    return arr, comparisons, movements, (end_time - start_time)


def run_selection_sort(input_array):
    arr = list(input_array)  
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        if min_idx != i:
            
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            movements += 3  
    end_time = time.perf_counter()
    
    return arr, comparisons, movements, (end_time - start_time)


def run_insertion_sort(input_array):
    arr = list(input_array)  
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        movements += 1  
        j = i - 1
        
        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1  
                j -= 1
            else:
                break
                
        arr[j + 1] = key
        movements += 1  
    end_time = time.perf_counter()
    
    return arr, comparisons, movements, (end_time - start_time)

print("Sorting algorithms loaded and configured with tracking monitors.")
```

```python

conditions = {
    "Sorted": sorted_input,
    "Reverse": reverse_input,
    "Random": random_input
}

benchmark_results = []

def verify_sorted(arr):
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            return "FAILED"
    return "VERIFIED PASSED"

for cond_name, datasets in conditions.items():
    b_arr, b_comp, b_mov, b_time = run_bubble_sort(datasets)
    benchmark_results.append({"Algo": "Bubble Sort", "Condition": cond_name, "Comp": b_comp, "Mov": b_mov, "Time": b_time, "Status": verify_sorted(b_arr)})

    s_arr, s_comp, s_mov, s_time = run_selection_sort(datasets)
    benchmark_results.append({"Algo": "Selection Sort", "Condition": cond_name, "Comp": s_comp, "Mov": s_mov, "Time": s_time, "Status": verify_sorted(s_arr)})
 
    i_arr, i_comp, i_mov, i_time = run_insertion_sort(datasets)
    benchmark_results.append({"Algo": "Insertion Sort", "Condition": cond_name, "Comp": i_comp, "Mov": i_mov, "Time": i_time, "Status": verify_sorted(i_arr)})

print("Testing and Comparison Completed.")
for res in benchmark_results:
    print(f"{res['Algo']:<15} | Condition: {res['Condition']:<8} | Status: {res['Status']}")
```

```python

print("\n1. Bubble Sort Matrix:")
print("   - Best Case (Pre-sorted):   O(n) if optimized, but standard runs O(n²)")
print("   - Average / Worst Case:     O(n²) - High comparisons and endless swaps.")

print("\n2. Selection Sort Matrix:")
print("   - All Cases (Any order):    O(n²) - Constantly scans the remaining space,")
print("                                 regardless of whether it's pre-sorted or not.")

print("\n3. Insertion Sort Matrix:")
print("   - Best Case (Pre-sorted):   O(n) - Fast linear scans with no data shifting.")
print("   - Average / Worst Case:     O(n²) - Deep shifting penalties on reversed data.")
```
